```python
import matplotlib.pyplot as plt

df['ChurnLabel'].value_counts().plot(kind='bar')

plt.title('Customer Churn Distribution')
plt.xlabel('Churn Status')
plt.ylabel('Number of Customers')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

contract_churn = pd.crosstab(df['Contract'], df['ChurnLabel'])

contract_churn.plot(kind='bar', figsize=(8,5))

plt.title('Contract Type vs Customer Churn')
plt.xlabel('Contract Type')
plt.ylabel('Number of Customers')
plt.legend(title='Churn')
plt.show()
```

```python
internet_churn = pd.crosstab(df['InternetType'], df['ChurnLabel'])

internet_churn.plot(kind='bar', figsize=(8,5))

plt.title('Internet Type vs Customer Churn')
plt.xlabel('Internet Type')
plt.ylabel('Number of Customers')
plt.legend(title='Churn')

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df['MonthlyCharge'], bins=20)

plt.title('Monthly Charges Distribution')
plt.xlabel('Monthly Charges')
plt.ylabel('Number of Customers')

plt.show()
```

```python
import matplotlib.pyplot as plt

df.boxplot(column='MonthlyCharge', by='ChurnLabel', figsize=(8,5))

plt.title('Monthly Charges vs Churn')
plt.suptitle('')
plt.xlabel('Churn Status')
plt.ylabel('Monthly Charges')

plt.show()
```

```python
import matplotlib.pyplot as plt

df['ChurnReason'].value_counts().head(10).plot(
    kind='barh',
    figsize=(10,6)
)

plt.title('Top 10 Churn Reasons')
plt.xlabel('Number of Customers')
plt.ylabel('Churn Reason')

plt.show()
```

```python
print("Dataset Shape:", df.shape)

print("\nTotal Customers:", len(df))

print("\nChurn Distribution:")
print(df['ChurnLabel'].value_counts())

print("\nContract Types:")
print(df['Contract'].value_counts())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

satisfaction_churn = pd.crosstab(df['SatisfactionScore'], df['ChurnLabel'])

satisfaction_churn.plot(kind='bar', figsize=(10,5))

plt.title('Satisfaction Score vs Customer Churn')
plt.xlabel('Satisfaction Score')
plt.ylabel('Number of Customers')
plt.legend(title='Churn')

plt.show()
```

```python
df['CustomerStatus'].value_counts().plot(kind='bar')

plt.title('Customer Status Distribution')
plt.xlabel('Customer Status')
plt.ylabel('Number of Customers')

plt.show()
```

```python
churn_rate = (df['ChurnLabel'].value_counts()['Yes'] / len(df)) * 100

print(f"Customer Churn Rate: {churn_rate:.2f}%")
```

# Customer Churn Analysis

## Objective
The goal of this project is to analyze customer churn behavior and identify the key factors influencing customer retention and customer loss.

## Dataset
IBM Telco Customer Churn Dataset

## Tools Used
- Python
- Pandas
- Matplotlib
- Google Colab

## Key Findings
- Customer Churn Rate: 26.54%
- Month-to-Month contracts show the highest churn.
- Fiber Optic customers exhibit higher churn compared to other internet types.
- Customers with low satisfaction scores are more likely to churn.
- Competitor offers and better devices are major churn reasons.

## Conclusion
Customer satisfaction, contract type, internet service type, and competitor offerings significantly influence customer churn. Businesses can improve retention by enhancing customer experience and offering competitive plans.
